```python
import os, subprocess

OUT = "fx_data"
start, end = "2026-01-01", "2026-01-10"
symbols = ["USDBRL"]

# 1) create folder
os.makedirs(OUT, exist_ok=True)
print("created:", os.path.exists(OUT))

# 2) run duka
subprocess.run(
    ["duka", *symbols, "-s", start, "-e", end, "-c", "m1", "-f", OUT],
    check=True
)

print("done")
```
